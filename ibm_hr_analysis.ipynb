```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# settings
sns.set()
```

```python
df = pd.read_csv("C:\\Users\\palak_1816\\Downloads\\hr_data.csv")  

df.head()
```

```python
df.shape
df.info()
df.describe()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
df.drop_duplicates(inplace=True)
```

```python
sns.countplot(x='Attrition', data=df)
plt.title("Attrition Count")
plt.show()

df['Attrition'].value_counts(normalize=True) * 100
```

```python
sns.histplot(df['Age'], bins=20, kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
sns.countplot(x='Department', hue='Attrition', data=df)
plt.title("Attrition by Department")
plt.xticks(rotation=30)
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x='JobRole', hue='Attrition', data=df)
plt.xticks(rotation=45)
plt.title("Attrition by Job Role")
plt.show()
```

```python
sns.boxplot(x='Attrition', y='MonthlyIncome', data=df)
plt.title("Salary vs Attrition")
plt.show()
```

```python
sns.countplot(x='OverTime', hue='Attrition', data=df)
plt.title("Overtime vs Attrition")
plt.show()
```

```python
sns.countplot(x='WorkLifeBalance', hue='Attrition', data=df)
plt.title("Work-Life Balance vs Attrition")
plt.show()
```

```python
plt.figure(figsize=(12,8))

corr = df.corr(numeric_only=True)

sns.heatmap(corr, 
            cmap='coolwarm', 
            annot=False,   # ❌ remove numbers (main reason messy)
            linewidths=0.5)

plt.title("Correlation Heatmap", fontsize=14)
plt.xticks(rotation=45)
plt.yticks(rotation=0)

plt.show()
```

```python
bins = [18, 30, 40, 60]
labels = ['20-30', '30-40', '40+']

df['AgeGroup'] = pd.cut(df['Age'], bins=bins, labels=labels)

sns.countplot(x='AgeGroup', hue='Attrition', data=df)
plt.title("Attrition by Age Group")
plt.show()
```

Key Insights:
- Higher attrition seen in employees working overtime
- Lower salary group has more attrition
- Certain departments (like Sales) show higher attrition
- Younger employees tend to leave more
